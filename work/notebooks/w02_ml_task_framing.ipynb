## 1. My lane as an ML task (type)

## My lane as an ML task

My lane is **Refresh / Content Opportunity Scoring**, and I frame it primarily as a **ranking problem**. The goal is to rank content pages by their priority for human review, so a content team can focus its limited time on the pages that appear most worth reviewing first.

Ranking is appropriate because the final output is not simply a yes/no decision. Instead, pages need to be ordered from higher to lower review priority using multiple observable search and engagement signals. The ranking can support actions such as refreshing, improving, or monitoring a page. The output is decision-support, not a guarantee that a page will improve after an action.

## Target or proxy

For the starter version of this task, I will use `trend_direction` as a proxy target. A page with `trend_direction == "down"` is treated as an observed declining page in the current data window.

This is a defined proxy based on an observed field, not a future outcome. ## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success metric

My primary success metric will be **Precision@50**. This measures how many of the top 50 pages ranked for review match the defined target or proxy.

Precision@50 fits the real decision because a content team has limited review capacity and may only be able to inspect a fixed number of pages first. A useful ranking should therefore put more relevant pages near the top rather than simply classifying every page correctly.

For the starter proxy, a higher Precision@50 means that the ranking is better at putting observed declining pages among the pages recommended for early review. This metric measures prioritization quality, not whether a refresh will cause a page to recover.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML beats a fixed rule here

A fixed rule can identify pages using one or a few manually chosen thresholds, but it may not capture the different combinations of search and engagement signals that make one page a higher-priority review candidate than another.

For example, two pages may both have a declining trend, but they can differ substantially in impressions, sessions, CTR, average position, content age, and engagement. A fixed rule would require manually setting many thresholds and deciding how these signals interact.

Machine learning can learn patterns across multiple observable signals and produce a ranked prioritization that can be compared with a transparent baseline rule. The purpose is not to assume that ML is automatically better, but to test whether it improves Precision@50 and provides more useful decision-support for content review.
